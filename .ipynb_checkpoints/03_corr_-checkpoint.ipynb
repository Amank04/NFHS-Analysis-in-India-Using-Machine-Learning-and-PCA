```python
import pandas as pd

# Step 1: Fill NaN values with column means
x1 = df.drop([0, 2, 3], axis=1)
x1 = x1.apply(lambda col: col.fillna(col.mean()), axis=0)

# Step 2: Calculate the correlation matrix
correlation_matrix = x1.corr()

# Step 3: Define the correlation threshold for multicollinearity
threshold = 0.5

# Step 4: Initialize a dictionary to store output feature sets
high_correlation_sets = {}

# Step 5: Function to remove highly correlated features
def remove_multicollinearity(features, corr_matrix, threshold):
    removed_features = []
    for feature in features:
        # Check the correlation of this feature with others in the set
        for other_feature in features:
            if feature != other_feature:
                if abs(corr_matrix[feature][other_feature]) >= threshold:
                    removed_features.append(other_feature)
                    break  # Remove only one at a time to avoid over-removing features
    return [f for f in features if f not in removed_features]

# Step 6: Iterate over each column in the correlation matrix
for col in correlation_matrix.columns:
    # Find features with high correlation to the current column
    correlated_features = correlation_matrix.index[correlation_matrix[col].abs() >= threshold].tolist()
    
    # Remove the current column from its own list of correlated features
    correlated_features.remove(col)
    
    # If there are correlated features, process them
    if correlated_features:
        filtered_features = remove_multicollinearity(correlated_features, correlation_matrix, threshold)
        high_correlation_sets[col] = filtered_features

# Step 7: Create a DataFrame to store output and input features
feature_sets = []
for output, inputs in high_correlation_sets.items():
    feature_sets.append({'Output Feature': output, 'Input Features': inputs})

feature_df = pd.DataFrame(feature_sets)

# Step 8: Save the DataFrame to a CSV file for future editing
feature_df.to_csv('high_correlation_feature_sets.csv', index=False)

# Optional: Display the output and corresponding input feature sets
pd.options.display.max_rows = 100
print(feature_df)
```

#PCA (not useful right now)

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Initialize a dictionary to store the reduced features after PCA
pca_reduced_features = {}

# Iterate over each output feature and its corresponding correlated input features
for output, inputs in high_correlation_sets.items():
    # Extract the input feature set from the original DataFrame
    input_data = x1[inputs]
    
    # Check if the input_data is empty and continue if it is
    if input_data.empty:
        print(f"Skipping {output} as input data is empty.")
        continue

    # Display the shape of input data for debugging
    print(f"Input data shape for output {output}: {input_data.shape}")
    
    # Standardize the input data (PCA is sensitive to scale)
    scaler = StandardScaler()
    input_data_scaled = scaler.fit_transform(input_data)
    
    # Apply PCA
    pca = PCA()
    pca_components = pca.fit_transform(input_data_scaled)
    
    # Choose the number of components that explain at least 95% of the variance
    explained_variance_ratio = pca.explained_variance_ratio_
    cumulative_variance = explained_variance_ratio.cumsum()
    n_components = (cumulative_variance >= 0.95).argmax() + 1
    
    # Transform the data using the selected number of components
    pca_reduced = pca_components[:, :n_components]
    
    # Store the reduced features in the dictionary
    pca_reduced_features[output] = pca_reduced

    # Display the result
    print(f"Output Feature: {output}")
    print(f"Number of PCA components retained: {n_components}")
    print(f"Reduced input features:\n{pca_reduced}\n")
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LinearRegression

# Fill NaN values with column means (as you've already done)
x1 = df.drop([0, 2, 3], axis=1)
x1 = x1.apply(lambda col: col.fillna(col.mean()), axis=0)

# Calculate the correlation matrix
correlation_matrix = x1.corr()

# Define the correlation threshold
threshold = 0.8

# Initialize a dictionary to store feature sets (key = output feature, value = list of input features)
high_correlation_sets = {}

# Iterate over each column in the correlation matrix
for col in correlation_matrix.columns:
    # Find the features where the correlation with the current column is >= threshold
    correlated_features = correlation_matrix.index[correlation_matrix[col].abs() >= threshold].tolist()
    
    # Remove the current column from its own list of correlated features (because it will always be perfectly correlated with itself)
    correlated_features.remove(col)
    
    # If there are any other correlated features, add them as input features for the current column (output)
    if correlated_features:
        high_correlation_sets[col] = correlated_features

# Initialize a linear regression model
model = LinearRegression()

# Iterate over each set in high_correlation_sets and perform cross-validation
for output, inputs in high_correlation_sets.items():
    # Limit to sets with only 1 or 2 input features
    if len(inputs) <= 2 :
        X = x1[inputs]  # Input features
        y = x1[output]  # Output feature
        
        # Perform cross-validation and compute the mean score (R²)
        scores = cross_val_score(model, X, y, cv=5, scoring='r2')  # 5-fold cross-validation
        mean_score = np.mean(scores)

        # Print the result for each set
        print(f"Output Feature: {output}")
        print(f"Input Features: {inputs}")
        print(f"Mean Cross-Validation R²: {mean_score}\n")
```
