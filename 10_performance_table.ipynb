```python
import pandas as pd

# Load the results CSV file
results_df = pd.read_csv('modified_regression_results.csv')

# Create the pivot table
pivot_table = results_df.pivot_table(
    index='Output Feature', 
    columns='Model', 
    values=['RMSE', 'R2 Score'],
    aggfunc='mean'
)

# Extract RMSE and R² separately
rmse_values = pivot_table['RMSE']
r2_values = pivot_table['R2 Score']

# Save RMSE and R² values to LaTeX format
rmse_values.to_latex('rmse_values.tex', header=True, index=True)
r2_values.to_latex('r2_values.tex', header=True, index=True)

print("RMSE and R² values saved in LaTeX format.")
```

```python
import pandas as pd

# Load the results CSV file
results_df = pd.read_csv('regression_results.csv')

# Create the pivot table for better readability
pivot_table = results_df.pivot_table(
    index='Output Feature', 
    columns='Model', 
    values=['RMSE', 'R2 Score'],
    aggfunc='mean'
)

# Generate LaTeX table from the pivot table
latex_table = '\\begin{table}[htbp]\n\\centering\n\\caption{Performance Comparison of Machine Learning Models}\n\\begin{tabular}{|l|' + '|'.join(['c'] * len(pivot_table.columns)) + '|}\n\\hline\n'
latex_table += 'Model & ' + ' & '.join(pivot_table.columns.get_level_values(1).unique()) + ' \\\\ \\hline\n'

# Loop through rows (Output Features)
for row in pivot_table.index:
    # Ensure row name is string and add the corresponding RMSE and R² values for each model
    row_str = str(row)  # Convert row name to string
    values = ' & '.join([f'{pivot_table.at[row, (metric, model)]:.2f}' for metric, model in pivot_table.columns])
    latex_table += f"{row_str} & {values} \\\\ \\hline\n"

latex_table += '\\end{tabular}\n\\end{table}'

print(latex_table)
```
