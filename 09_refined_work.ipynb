### Weights to ohd indicators(District) without IMR,Hospitals & Life Expectency.

```python
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler


df = pd.read_csv('district_ohd_indicators.csv')
# df["2"]

# Separate the district column and the indicators
districts = df["2"]
indicators = df.drop(columns=["2",'index','4','35','22','19','81'])
df = df.drop(columns=['4','35','22','19','81'])

# Step 1: Standardize the indicator columns
scaler = StandardScaler()
indicators_scaled = scaler.fit_transform(indicators)

# Step 2: Run PCA and get the first principal component
pca = PCA(n_components=1)
pca.fit(indicators_scaled)

# Extract the loadings (weights) of each indicator in the first principal component
pc1_loadings = pca.components_[0]

# Step 3: Convert loadings to absolute values and normalize to get weights
abs_loadings = abs(pc1_loadings)
weights = abs_loadings / abs_loadings.sum()

# Create a DataFrame to display weights for each indicator
weights_df = pd.DataFrame({
    'indicator': indicators.columns,
    'weight': weights
})

# Display the weights for each indicator
print("Weights for each indicator:")
print(weights_df)

# Step 4: Calculate the overall health condition score for each district
# Multiply each indicator by its weight and sum across columns
health_scores = indicators_scaled.dot(weights)

# Step 5: Add the health condition score to the original DataFrame
df['overall_health_condition'] = health_scores

# Display the updated DataFrame with the overall health condition score
print("\nDataFrame with Overall Health Condition Score:")
print(df[["2", 'overall_health_condition']])

df = df.drop(columns=['index'])
df.to_csv("PCA_No_IMR_health_conditions_by_district.csv")
```

## Fuzzymatching

```python
import pandas as pd
from rapidfuzz import process, fuzz


# Convert to DataFrames
df1 = pd.read_csv("PCA_No_IMR_health_conditions_by_district.csv")
df2 = pd.read_csv("Geojson_District_mapping.csv")

# Columns to match
column1 = '2'
column2 = '2'

# Function to handle user decision
def match_and_decide(row, choices):
    match, score, idx = process.extractOne(
        row[column1], 
        choices, 
        scorer=fuzz.token_sort_ratio
    )
    if score >= 70:  # Define a threshold for similarity
        print(f"Potential match found for '{row[column1]}': '{match}' (Score: {score})")
        user_input = input("Do you want to replace it? (yes/no): ").strip().lower()
        if user_input in ['yes', 'y']:
            return match
    return row[column1]

# Perform fuzzy matching and update based on user input
df1['Matched District'] = df1.apply(
    lambda row: match_and_decide(row, df2[column2].tolist()), axis=1
)

# Show updated DataFrame
print("\nFinal DataFrame:")
print(df1)
```
