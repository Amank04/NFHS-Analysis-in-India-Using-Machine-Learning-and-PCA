### Overall health Condition

```python
df = df[[2,73, 74, 39, 4, 33, 34, 35, 22, 19, 80, 81, 97]]
df.to_csv("ohd_indicators.csv")
```

```python
import pandas as pd

# Assuming you already have your DataFrame `df` loaded with 733 rows and the appropriate columns
# Example: Loading data (you might already have your data loaded in `df`)
# df = pd.read_csv('your_file.csv')  # Example for loading from a CSV

# Define a scoring function to calculate health score
def calculate_health_score(row):
    score = (
        row[73] * 0.15 +  # Mothers who had at least 4 antenatal care visits
    row[74] * 0.15 +  # Mothers who received postnatal care
    row[39] * 0.10 +  # Children who received postnatal care
    (100 - row[4]) * 0.10 +  # Women age 15-19 who are anemic (inverse)
    (100 - row[33]) * 0.15 +  # Children under 5 who are stunted (inverse)
    (100 - row[34]) * 0.10 +  # Children under 5 who are underweight (inverse)
    (100 - row[35]) * 0.10 +  # Children under 5 who are wasted (inverse)
    row[22] * 0.05 +  # Children age 12-23 months who received measles vaccine
    row[19] * 0.05 +  # Children age 12-23 months who received polio vaccine
    row[80] * 0.05 +  # Population using improved sanitation facility
    row[81] * 0.05 +  # Population using improved drinking water source
    row[97] * 0.05    # Women age 15-49 who are literate

    )
    return score

# Calculate health score for all rows in the DataFrame using .loc to avoid warning
df.loc[:, 'health_score'] = df.apply(calculate_health_score, axis=1)

# Classify overall health condition based on health score
def classify_health_condition(score):
    if score > 85:
        return 'Good'
    elif score > 60:
        return 'Moderate'
    else:
        return 'Poor'

# Apply classification to all health scores and add a new column for overall health condition
df.loc[:, 'overall_health_condition'] = df['health_score'].apply(classify_health_condition)


# Display the entire DataFrame (optional: you can limit the rows displayed)
df

# Optionally, if you want to see a limited number of rows for readability
# print(df.head(10))  # Display the first 10 rows

# Optionally, save the DataFrame to a new CSV file
df.to_csv('health_conditions_by_district.csv', index=False)
```

### Weights to ohd indicators

```python
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Example DataFrame structure
# df = pd.DataFrame({
#     'district': [...],
#     'indicator_1': [...],
#     'indicator_2': [...],
#     ...
# })

df = pd.read_csv('ohd_indicators.csv')
# df["2"]

# Separate the district column and the indicators
districts = df["2"]
indicators = df.drop(columns=["2",'index'])

# Step 1: Standardize the indicator columns
scaler = StandardScaler()
indicators_scaled = scaler.fit_transform(indicators)

# Step 2: Run PCA and get the first principal component
pca = PCA(n_components=1)
pca.fit(indicators_scaled)

# Extract the loadings (weights) of each indicator in the first principal component
pc1_loadings = pca.components_[0]

# Step 3: Convert loadings to absolute values and normalize to get weights
abs_loadings = abs(pc1_loadings)
weights = abs_loadings / abs_loadings.sum()

# Create a DataFrame to display weights for each indicator
weights_df = pd.DataFrame({
    'indicator': indicators.columns,
    'weight': weights
})

# Display the weights for each indicator
print("Weights for each indicator:")
print(weights_df)

# Step 4: Calculate the overall health condition score for each district
# Multiply each indicator by its weight and sum across columns
health_scores = indicators_scaled.dot(weights)

# Step 5: Add the health condition score to the original DataFrame
df['overall_health_condition'] = health_scores

# Display the updated DataFrame with the overall health condition score
print("\nDataFrame with Overall Health Condition Score:")
print(df[["2", 'overall_health_condition']])
```
