```python
import pandas as pd

# Load dataset
data = pd.read_csv('NHS-5_2019-20.csv')

# Filter out non-numeric columns
numeric_data = data.select_dtypes(include=[float, int])

numeric_data = numeric_data.apply(lambda col: col.fillna(col.mean()), axis=0)
# Calculate Spearman Correlation
spearman_corr = numeric_data.corr(method='spearman')

# Filter for absolute correlation greater than or equal to 0.7
spearman_significant = spearman_corr[abs(spearman_corr) >= 0.7]
spearman_significant = spearman_significant[spearman_significant != 1]  # Remove self-correlations

# Print Spearman significant correlations
print("Significant Spearman Correlations (|r| >= 0.7):")
print(spearman_significant.dropna(how='all'))  # Drop rows with all NaN values

# Calculate Kendall Correlation
kendall_corr = numeric_data.corr(method='kendall')

# Filter for absolute correlation greater than or equal to 0.7
kendall_significant = kendall_corr[abs(kendall_corr) >= 0.7]
kendall_significant = kendall_significant[kendall_significant != 1]  # Remove self-correlations

# Print Kendall significant correlations
print("\nSignificant Kendall Correlations (|r| >= 0.7):")
print(kendall_significant.dropna(how='all'))  # Drop rows with all NaN values
```
